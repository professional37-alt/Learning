```python
# Langrange interpolation 
import numpy as np
x = np.array([2,4,6,8,10])
y = np.array([1,3,5,7,9])

def Lagrng(x,y,xx):
    n = len(x)
    r = 0
    for i in range(n):
         s = y[i]
         for j in range(n):
             if i != j:
                 s = s*((xx - x[j]) / (x[i] - x[j]))
         r = r + s
    return r

xx = 3
yy = Lagrng(x, y, xx)


# Generate coordinates for plotting the polynomial
x_plot = np.linspace(2, 10, 300)
y_plot = np.array([Lagrng(x, y, z) for z in x_plot])

# Plot
plt.figure(figsize=(9, 5))

plt.plot(x_plot, y_plot, label="Lagrange interpolation")
plt.scatter(x, y, color="blue", s=60, label="Known data points")
plt.scatter(xx, yy, color="red", s=60, label=f"Interpolated point ({xx}, {yy})")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Lagrange Interpolation")
plt.grid(True)
plt.legend()
plt.show()

print("Interpolated value at x =", xx, "is", yy)
     
```

```python
# random number generator for NACA 4-digit airfoil
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng()

m = rng.choice([0.00, 0.01, 0.02, 0.03, 0.04])
p = rng.choice([0.2, 0.3, 0.4, 0.5])
t = rng.choice([0.08, 0.10, 0.12, 0.15])

M = int(round(m * 100))
P = int(round(p * 10))
T = int(round(t * 100))


print("NACA airfoil: NACA {}{}{:02d}".format(M, P, T))
print("Maximum camber  c =", m)
print("Location of maximum camber x =", p)
print("Maximum thickness ratio t =", t)
  

       ###################################################################


# Chordwise locations
N = 101
beta = np.linspace(0, np.pi, N)
x = (1 - np.cos(beta)) / 2

# Thickness distribution
yt = 5 * t * (
    0.2969 * np.sqrt(x)
    - 0.1260 * x
    - 0.3516 * x**2
    + 0.2843 * x**3
    - 0.1015 * x**4)


# Mean camber line and its slope
yc = np.zeros(N)
dyc = np.zeros(N)

for i in range(N):
    if x[i] <= p:
        yc[i] = (m / p**2) * (2*p*x[i] - x[i]**2)
        dyc[i] = (2*m / p**2) * (p - x[i])
    else:
        yc[i] = (m / (1-p)**2) * (
            (1 - 2*p) + 2*p*x[i] - x[i]**2)
        dyc[i] = (2*m / (1-p)**2) * (p - x[i])

theta = np.arctan(dyc)


       ###################################################################


xu = x - yt * np.sin(theta)
yu = yc + yt * np.cos(theta)

xl = x + yt * np.sin(theta)
yl = yc - yt * np.cos(theta)

plt.figure(figsize=(10, 4))

plt.plot(xu, yu, label="Upper surface", color="blue")
plt.plot(xl, yl, label="Lower surface", color="green")
plt.plot(x, yc, "--", label="Mean Camber Line", color="red")

plt.xlabel("x / c")
plt.ylabel("y / c")
plt.title("Original NACA {}{}{:02d}".format(M, P, T))
plt.axis("equal")
plt.grid(True)
plt.legend()
plt.show()


         ###################################################################


# Select six points from each surface
N_points = 6
indices = np.linspace(0, N-1, N_points, dtype=int)

x_upper = xu[indices]
y_upper = yu[indices]

x_lower = xl[indices]
y_lower = yl[indices]

# Plot the original airfoil and the six selected points
plt.figure(figsize=(10, 5))

# Six selected points on each surface
plt.scatter(x_upper, y_upper,color="red", s=60,label="6 selected upper points")
plt.scatter(x_lower, y_lower,color="blue", s=60,label="6 selected lower points")

plt.xlabel("x/c")
plt.ylabel("y/c")
plt.title("Six Selected Points on the NACA Airfoil")
plt.axis("equal")
plt.grid(True)
plt.legend()
plt.show()
```
