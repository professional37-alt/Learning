```python
import numpy as np
x = np.array([2,4,6,8,10,12])
y = np.array([1,3,5,7,9,11])
xx = 4

n = len(x)
r = 0
for i in range(n):
    s = y[i]
    for j in range(n):
        if i != j:
            s = s*((xx - x[j]) / (x[i] - x[j]))
    r = r + s
print(r)            
            
```

```python
import numpy as np
x = np.array([2,4,6,8,10])
y = np.array([1,3,5,7,9])
xx = 6

def LagInt(x,y,xx):
    n = len(x)
    r = 0
    for i in range(n):
         s = y[i]
         for j in range(n):
             if i != j:
                 s = s*((xx - x[j]) / (x[i] - x[j]))
         r = r + s
    return r
print(LagInt(x,y,xx))    
     
```
