```python
import numpy as np
x = np.array([2,4,6,8,10,12])
y = np.array([1,3,5,7,9,11])
xx = 12

n = len(x)
r = 0
for i in range(n):
    s = y[i]
    for j in range(n):
        if i != j:
            s = s*((xx - x[j]) / (x[i] - x[j]))
    r = r + s
print(r)            
            
```

```python
# Langrange interpolation 
import numpy as np
x = np.array([2,4,6,8,10])
y = np.array([1,3,5,7,9])
xx = 3

def Lagrng(x,y,xx):
    n = len(x)
    r = 0
    for i in range(n):
         s = y[i]
         for j in range(n):
             if i != j:
                 s = s*((xx - x[j]) / (x[i] - x[j]))
         r = r + s
    return r
print(Lagrng(x,y,xx))    
     
```

```python
# random number generator for NACA 4-digit airfoil
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng()

m = rng.choice([0.00, 0.01, 0.02, 0.03, 0.04])
p = rng.choice([0.2, 0.3, 0.4, 0.5])
t = rng.choice([0.08, 0.10, 0.12, 0.15])

M = int(round(m * 100))
P = int(round(p * 10))
T = int(round(t * 100))


print("NACA airfoil: NACA {}{}{:02d}".format(M, P, T))
print("Maximum camber  c =", m)
print("Location of maximum camber x =", p)
print("Maximum thickness ratio t =", t)

       ###################################################################


# Chordwise locations
N = 101
beta = np.linspace(0, np.pi, N)
x = (1 - np.cos(beta)) / 2

# Thickness distribution
yt = 5 * t * (
    0.2969 * np.sqrt(x)
    - 0.1260 * x
    - 0.3516 * x**2
    + 0.2843 * x**3
    - 0.1015 * x**4)


# Mean camber line and its slope
yc = np.zeros(N)
dyc = np.zeros(N)

for i in range(N):
    if x[i] <= p:
        yc[i] = (m / p**2) * (2*p*x[i] - x[i]**2)
        dyc[i] = (2*m / p**2) * (p - x[i])
    else:
        yc[i] = (m / (1-p)**2) * (
            (1 - 2*p) + 2*p*x[i] - x[i]**2)
        dyc[i] = (2*m / (1-p)**2) * (p - x[i])

theta = np.arctan(dyc)

print("mean camber height = ", yc)
print("slope of mean camber line = ", dyc)
print("angle btw mean camber and chord line = ", theta)

        ###################################################################



xu = x - yt * np.sin(theta)
yu = yc + yt * np.cos(theta)

xl = x + yt * np.sin(theta)
yl = yc - yt * np.cos(theta)

print("upper x points =", xu)
print("upper y points =", yu)
print("lower x points =", xl)
print("lower y points =", yl)
```
